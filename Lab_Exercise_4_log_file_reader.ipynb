 1.Develop a python program to read a log file named log.txt.

```python
%%writefile log.txt
10:00 INFO Application started
10:05 WARNING Low memory
10:10 ERROR Database connection failed
10:15 INFO User logged in
10:20 ERROR File not found
10:25 INFO Application closed
```

```python
# Lab Exercise 1: Log File Reader Program

try:
    # Open the log file in read mode
    with open("log.txt", "r") as file:

        error_count = 0

        print("Log File Contents:")
        print("------------------")

        # Read the file line by line
        for line in file:
            print(line.strip())

            # Count ERROR messages
            if "ERROR" in line:
                error_count += 1

        # Display total errors
        print("------------------")
        print("Total number of errors:", error_count)

except FileNotFoundError:
    print("Error: Log file does not exist!")
```

2.Develop a python program to manage student records using a text file named students.txt.

```python
# Lab Exercise 2: Student Data Manager Using Files

def add_student():
    roll = input("Enter Roll Number: ")
    name = input("Enter Student Name: ")
    marks = input("Enter Marks: ")

    with open("students.txt", "a") as file:
        file.write(roll + "," + name + "," + marks + "\n")

    print("Student record added successfully!")


def display_students():
    try:
        with open("students.txt", "r") as file:
            print("\nStudent Records:")
            print("--------------------")

            for line in file:
                roll, name, marks = line.strip().split(",")
                print("Roll Number:", roll)
                print("Student Name:", name)
                print("Marks:", marks)
                print("--------------------")

    except FileNotFoundError:
        print("No student records found!")


while True:
    print("\n1. Add Student")
    print("2. Display Students")
    print("3. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        add_student()

    elif choice == "2":
        display_students()

    elif choice == "3":
        print("Exiting the program...")
        break

    else:
        print("Invalid choice! Try again.")
```

3.Develop a python program to create a simple payroll management system using the csv module

```python
# Lab Exercise 3: CSV-Based Payroll Management System

import csv

try:
    n = int(input("Enter number of employees: "))

    # Create and write CSV file
    with open("payroll.csv", "w", newline="") as file:
        writer = csv.writer(file)

        writer.writerow(["Employee ID", "Name", "Basic Salary",
                         "Allowance", "Deduction", "Net Salary"])

        for i in range(n):
            print("\nEnter details of Employee", i + 1)

            emp_id = input("Enter Employee ID: ")
            name = input("Enter Name: ")
            basic = float(input("Enter Basic Salary: "))
            allowance = float(input("Enter Allowance: "))
            deduction = float(input("Enter Deduction: "))

            net_salary = basic + allowance - deduction

            writer.writerow([emp_id, name, basic,
                             allowance, deduction, net_salary])

    print("\nPayroll records saved successfully!")

    # Read and display CSV file
    print("\nEmployee Payroll Records:")
    print("--------------------------------")

    with open("payroll.csv", "r") as file:
        reader = csv.reader(file)

        for row in reader:
            print(" | ".join(row))

except ValueError:
    print("Please enter valid numeric values!")
```

Develop a python application to analyse application logs stored in a JSON file named logs.json.

```python
%%writefile logs.json
[
    {
        "timestamp": "10:00",
        "level": "INFO",
        "message": "Application started"
    },
    {
        "timestamp": "10:05",
        "level": "WARNING",
        "message": "Low memory"
    },
    {
        "timestamp": "10:10",
        "level": "ERROR",
        "message": "Database connection failed"
    },
    {
        "timestamp": "10:15",
        "level": "INFO",
        "message": "User logged in"
    },
    {
        "timestamp": "10:20",
        "level": "ERROR",
        "message": "File not found"
    },
    {
        "timestamp": "10:25",
        "level": "INFO",
        "message": "Application closed"
    }
]
```

```python
# Mini Project: JSON-Based Log File Analyzer

import json

try:
    # Read JSON file
    with open("logs.json", "r") as file:
        logs = json.load(file)

    # Display all log records
    print("ALL LOG RECORDS")
    print("-------------------------")

    for log in logs:
        print("Timestamp:", log["timestamp"])
        print("Level:", log["level"])
        print("Message:", log["message"])
        print("-------------------------")

    # Count log categories
    info_count = 0
    warning_count = 0
    error_count = 0

    for log in logs:
        if log["level"] == "INFO":
            info_count += 1
        elif log["level"] == "WARNING":
            warning_count += 1
        elif log["level"] == "ERROR":
            error_count += 1

    # Display summary
    print("\nLOG SUMMARY")
    print("-------------------------")
    print("Total INFO messages:", info_count)
    print("Total WARNING messages:", warning_count)
    print("Total ERROR messages:", error_count)
    print("Total log records:", len(logs))

except FileNotFoundError:
    print("Error: JSON file not found!")

except json.JSONDecodeError:
    print("Error: Invalid JSON data!")
```
