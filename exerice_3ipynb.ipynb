```python
import pandas as pd

data = {
    "Name": ["Aman", "Rohit"],
    "Age": [20, 24]
}

df = pd.DataFrame(data)

print(df)
```

```python
# Write data into CSV file
df.to_csv("students.csv", index=False)

# Read data from CSV file
df_csv = pd.read_csv("students.csv")

print(df_csv)
```

```python
# Write data into JSON file
df.to_json("students.json", orient="records", indent=4)

# Read data from JSON file
df_json = pd.read_json("students.json")

print(df_json)
```

```python
import pandas as pd

# Creating data
data = {
    "Name": ["Aman", "Rohit"],
    "Age": [20, 24]
}

df = pd.DataFrame(data)

print("Original Data:")
print(df)


# CSV Operations
df.to_csv("students.csv", index=False)

df_csv = pd.read_csv("students.csv")

print("\nCSV Data:")
print(df_csv)


# JSON Operations
df.to_json("students.json", orient="records", indent=4)

df_json = pd.read_json("students.json")

print("\nJSON Data:")
print(df_json)
```
