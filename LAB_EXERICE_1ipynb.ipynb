Objective
Write a Python program that accepts a student's personal and academic details and displays them in a neatly formatted manner using formatted strings.


Problem Statement
Write a Python program to perform the following tasks:
1.Accept the following inputs from the user:
Student Name
USN
Branch
Semester
Three subject marks
2.Calculate:
Total Marks
Average Marks
3.Display all the information using formatted output.


```python
# Student Information and Academic Summary

name = input("Enter Student Name: ")
usn = input("Enter USN: ")
branch = input("Enter Branch: ")
semester = input("Enter Semester: ")

mark1 = float(input("Enter Marks in Subject 1: "))
mark2 = float(input("Enter Marks in Subject 2: "))
mark3 = float(input("Enter Marks in Subject 3: "))

total = mark1 + mark2 + mark3
average = total / 3

print("\nSTUDENT REPORT ")
print(f"Name       : {name}")
print(f"USN        : {usn}")
print(f"Branch     : {branch}")
print(f"Semester   : {semester}")
print(f"Total Marks: {total}")
print(f"Average    : {average:.2f}")
```

4. Title: Electricity Bill Generator


Objective
Develop a simple real-world billing application using Python variables, arithmetic operators, input/output statements, built-in functions, and formatted printing.


Problem Statement
A residential electricity provider wants a simple billing system.
Write a Python program that performs the following tasks:
Accept the following details from the user:
Consumer Name
Consumer ID
Previous Meter Reading (kWh)
Current Meter Reading (kWh)
Cost per Unit (₹)
Calculate:
Total Units Consumed
Energy Charge
Electricity Duty (5% of Energy Charge)
Fixed Meter Charge (₹100)
Net Bill Amount
Display the bill in a professional format.


```python
# Units Consumed = Current Reading − Previous Reading
# Energy Charge = Units   Cost per Unit
# Electricity Duty = 5% of Energy Charge
# Fixed Meter Charge = ₹100
# Net Bill = Energy Charge + Electricity Duty + Fixed Charge








# CODE:
print("========== ELECTRICITY BILL GENERATOR ==========")
consumer_name = input("Enter Consumer Name: ")
consumer_id = input("Enter Consumer ID: ")
previous_reading = float(input("Enter Previous Meter Reading (kWh): "))
current_reading = float(input("Enter Current Meter Reading (kWh): "))
cost_per_unit = float(input("Enter Cost per Unit (₹): "))
units = current_reading - previous_reading
energy_charge = units * cost_per_unit
electricity_duty = energy_charge * 0.05
fixed_charge = 100
net_bill = energy_charge + electricity_duty + fixed_charge

print("\n============== ELECTRICITY BILL ==============")
print(f"Consumer Name        : {consumer_name}")
print(f"Consumer ID          : {consumer_id}")
print(f"Units Consumed       : {units:.2f} kWh")
print(f"Energy Charge        : ₹{energy_charge:.2f}")
print(f"Electricity Duty(5%) : ₹{electricity_duty:.2f}")
print(f"Fixed Meter Charge   : ₹{fixed_charge:.2f}")
print("----------------------------------------------")
print(f"Net Bill Amount      : ₹{net_bill:.2f}")
print("==============================================")
```
